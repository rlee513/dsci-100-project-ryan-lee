#### Task 4.4: Plot the distribution of two relevant variables in the raw data (4 points)

```python
### EDIT THIS CODE CELL FOR PLOT 1 ###
import altair as alt
Plot1= alt.Chart(NCAA_Data).mark_bar().encode(
    x = alt.X("conf_wins_pct", bin=True).title("conference win percentage"),
    y= alt.Y("count()").title("number of appearances")).properties(title="distribution of conference win percentages for tournament teams")
Plot1
```

```python
### EDIT THIS CODE CELL FOR PLOT 2 ###
bid_props = NCAA_Data["bid"].value_counts(normalize=True).reset_index()
bid_props.columns = ["bid", "proportion"] 


Plot2 = alt.Chart(
    bid_props,
    title="Proportion of NCAA tournament appearances by bid type"
).mark_bar().encode(
    x=alt.X("bid").sort("-y").title("Type of tournament bid"),
    y=alt.Y("proportion").title("Proportion of tournament appearances")
)
Plot2
```

Visualization 1: This scatterplot shows the relationship between regional win pecentage and total win percentage with points coloured by bid type. Since my research questions asks whether win percentages can predict bid type, this plot shows whether two bid types occupy different regions of the prediction space. 

```python
### EDIT THIS CODE CELL FOR VISUALIZATION 1###
Vis1 = alt.Chart(
    NCAA_Data,
    title="Regional vs total win percentage by bid type").mark_point().encode(
    x=alt.X("reg_wins_pct").scale(zero=False).title("regional win percentage"),
    y=alt.Y("total_wins_pct").scale(zero=False).title("total win percentage"),
    color=alt.Color("bid")
        .title("Bid type"),
)

Vis1
```

Visualization 2: These histograms show the distribution of total win percentage seperately for auto and at-large bids. Comparing the centre and spread of each panel can show whether total win percentage differs between the two bid types which can be used  to see if it can help predict bid.

```python
### EDIT THIS CODE CELL FOR VISUALIZATION 2###
Vis2 = alt.Chart(NCAA_Data).mark_bar().encode(
    x=alt.X("total_wins_pct").bin().title("Total win percentage"),
    y=alt.Y("count()").title("Number of tournament appearances"),
    color=alt.Color("bid").title("Bid type"))

Vis2 = Vis2.properties(
    height=100
).facet(
    "bid:N",
    columns=1,
    title="Distribution of total win percentage by bid type")

Vis2
```
